```python
import numpy as np
import json
import glob
import os
repo_home = '../../'
```

```python
matt_folder = repo_home+'analyses/matt/'

json_files = glob.glob(os.path.join(matt_folder, "*.json"))

data = {}
for file in json_files:
    with open(file, 'r') as f:
        key = os.path.splitext(os.path.basename(file))[0]
        data[key] = json.load(f)
```

```python
data_new = {k:{} for k in data.keys()}

for k,v in data.items(): 
    print(k)
    for _k, _v in v.items():
        if 'ln_bayes_factor' in _k: 
            _k_new = _k.replace('ln', 'log10')
            _v_new = np.round( _v/np.log(10), 2)

            print(_k, _v )
            print(_k_new, _v_new)

            data_new[k][_k_new] = _v_new
```
